### Os tipos fundamentais

Quatro tipos básicos cobrem quase todas as nossas necessidades: inteiros (`int`), ponto flutuante (`float`), texto (`str`) e booleanos (`bool`). A função `type()` revela o tipo de qualquer valor.

### Aritmética e a unidade atômica de energia

A química quântica trabalha em **unidades atômicas**: a unidade de energia é o *hartree* (Ha). Guarde a conversão **1 Ha ≈ 27,2114 eV**.

```python
HARTREE_EM_EV = 27.2114        # constante de conversao

energia_H2_hartree = -1.11676  # energia da molecula H2 (Hartree-Fock/STO-3G)
energia_H2_eV = energia_H2_hartree * HARTREE_EM_EV

print("Energia da H2:", energia_H2_eV, "eV")
```

> ⚠️ **Atenção:** a divisão `/` sempre produz um `float` em Python 3 (`4 / 2` → `2.0`). Para divisão inteira use `//`.

## 1.3 Estruturas de dados: listas, tuplas e dicionários

### Listas — coleções ordenadas e modificáveis

A **indexação começa em zero**; índices negativos contam a partir do fim.

```python
# Energias dos orbitais moleculares (em hartree), do mais baixo ao mais alto
energias_orbitais = [-0.578, 0.670, 0.965, 1.410]

print("Numero de orbitais:", len(energias_orbitais))   # 4
print("Primeiro (indice 0):", energias_orbitais[0])    # -0.578
print("Ultimo (indice -1):", energias_orbitais[-1])    # 1.410
```

## 1.4 NumPy: o arsenal numérico

A mecânica quântica é, em essência, **álgebra linear**: estados são vetores, operadores são matrizes, energias são autovalores. O NumPy fornece o objeto `ndarray` e operações vetoriais rápidas. Por convenção, importamos com o apelido `np`.

```python
import numpy as np

# Um vetor de energias orbitais
energias = np.array([-0.578, 0.670, 0.965, 1.410])

# Uma matriz 2x2 (por exemplo, um Hamiltoniano simples)
H = np.array([[0.0, -1.0],
              [-1.0, 0.0]])

print("Formato do vetor:", energias.shape)   # (4,)
print("Formato da matriz:", H.shape)         # (2, 2)
```

### Operações vetorizadas

Operações se aplicam a todos os elementos de uma vez, sem laços explícitos.

```python
energias_hartree = np.array([-0.578, 0.670, 0.965, 1.410])

# Multiplica CADA elemento por 27.2114, de uma so vez
energias_eV = energias_hartree * 27.2114
print(energias_eV)

print("Energia minima:", energias_hartree.min())
print("Energia media:", energias_hartree.mean())
print("Soma das energias:", energias_hartree.sum())
```

### Álgebra linear: o coração da mecânica quântica

Considere o Hamiltoniano de dois níveis
$$\hat{H} = \begin{pmatrix} 0 & -t \\ -t & 0 \end{pmatrix},$$
que descreve um elétron que pode ocupar dois sítios acoplados por uma amplitude de tunelamento $t$. Seus **autovalores** são as energias permitidas.

> ⚠️ Para matrizes simétricas/Hermitianas (todo Hamiltoniano físico), use `np.linalg.eigh`, não `eig`: é mais rápida e devolve autovalores reais ordenados.

```python
t = 1.0
H = np.array([[0.0, -t],
              [-t,  0.0]])

# Para matrizes simetricas (Hermitianas), use eigh
autovalores, autovetores = np.linalg.eigh(H)

print("Energias (autovalores):", autovalores)   # [-1.  1.]
print("Estados (colunas):")
print(autovetores)
```

## 1.5 Controle de fluxo: condicionais e laços

### Condicionais — `if` / `else`

> ⚠️ **Indentação é sintaxe.** Python usa o recuo (4 espaços) para delimitar blocos — não há chaves.

```python
delta_energia = 3e-9   # variacao de energia no ultimo passo do SCF
limiar = 1e-8

if delta_energia < limiar:
    print("Convergiu! Podemos confiar no resultado.")
else:
    print("Ainda nao convergiu; aumente o numero de iteracoes.")
```

### O laço `for`

A função `enumerate()` devolve o índice **e** o valor. As *f-strings* (`f"..."`) inserem variáveis no texto com formatação controlada.

```python
energias_hartree = [-0.578, 0.670, 0.965, 1.410]

for i, e in enumerate(energias_hartree):
    e_eV = e * 27.2114
    print(f"Orbital {i}: {e:8.3f} Ha  =  {e_eV:8.2f} eV")
```

## 1.6 Funções e módulos

As **funções** encapsulam lógica sob um nome, permitindo reutilizá-la. Definimos com `def`; `return` entrega o resultado.

```python
def hartree_para_eV(energia_hartree):
    """Converte uma energia de hartree para eletron-volt."""
    return energia_hartree * 27.2114

print(hartree_para_eV(-0.5))      # -13.6057
print(hartree_para_eV(-1.11676))  # -30.39...
```

Funções podem ter argumentos com valores-padrão. Aqui reaproveitamos a diagonalização da seção anterior:

```python
def energia_ligante(t=1.0):
    """Energia do estado ligante de um sistema de dois sitios,
    acoplados por uma amplitude de tunelamento t (em hartree)."""
    H = np.array([[0.0, -t],
                  [-t,  0.0]])
    autovalores = np.linalg.eigh(H)[0]
    return autovalores[0]   # o menor autovalor: estado fundamental

print(energia_ligante())      # -1.0  (usa t = 1.0, o padrao)
print(energia_ligante(0.5))   # -0.5
```
